```python
write a function to check if the year number is leap year
```

```python
def is_leap_year(year):
    # Check if the year is a leap year
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False

# Take user input for the year
year = int(input("Enter a year: "))

# Check and print if the year is a leap year
if is_leap_year(year):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
write a function to take an array and return another array that contains the members of the first array that are even
```

```python
def get_even_numbers(arr):
    # Use list comprehension to filter even numbers
    even_numbers = [num for num in arr if num % 2 == 0]
    return even_numbers

# Example usage
input_array = list(map(int, input("Enter numbers separated by spaces: ").split()))
even_array = get_even_numbers(input_array)
print("Even numbers:", even_array)
```

```python
def get_even_numbers(arr):
    even_numbers = []
    for num in arr:
        if num % 2 == 0:
            even_numbers.append(num)
    return even_numbers

# Example usage
input_array = list(map(int, input("Enter numbers separated by spaces: ").split()))
even_array = get_even_numbers(input_array)
print("Even numbers:", even_array)
```

write a function that takes 2 array and prints the members of the first array that are present in the second array ( Hint: user membership comprehension)

```python
def common_members(arr1, arr2):
    # Use list comprehension to find common members
    common = [num for num in arr1 if num in arr2]
    return common

# Example usage
array1 = list(map(int, input("Enter numbers for the first array separated by spaces: ").split()))
array2 = list(map(int, input("Enter numbers for the second array separated by spaces: ").split()))

common_array = common_members(array1, array2)
print("Common members:", common_array)
```
