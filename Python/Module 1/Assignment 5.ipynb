```python
# Function to check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Take user-defined input to create an array
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Use a for loop to fetch and print prime numbers
for number in numbers:
    if is_prime(number):
        print(number)
```
