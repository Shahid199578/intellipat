```python
import pandas as pd

# Create the DataFrame
data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 5, 4, 2]
}
df = pd.DataFrame(data)

# Calculate the mean of columns x and y
mean_x = df['x'].mean()
mean_y = df['y'].mean()

# Calculate the mode of column x
mode_x = df['x'].mode()[0]

# Calculate the standard deviation of column y
std_dev_y = df['y'].std()

# Calculate the range of columns x and y
range_x = df['x'].max() - df['x'].min()
range_y = df['y'].max() - df['y'].min()

# Print the results
print(f"Mean of x: {mean_x}")
print(f"Mean of y: {mean_y}")
print(f"Mode of x: {mode_x}")
print(f"Standard Deviation of y: {std_dev_y}")
print(f"Range of x: {range_x}")
print(f"Range of y: {range_y}")
```
