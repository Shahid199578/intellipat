```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.cluster import KMeans

# Load the dataset
df = pd.read_csv("Live.csv")  # Update with the correct filename if needed

### 1) How many columns contain null data?
null_columns = df.isnull().sum()
num_null_cols = sum(null_columns > 0)
print(f"1) Columns with null data: {num_null_cols}")

### 2) How many status_ids are unique?
num_unique_status_ids = df["status_id"].nunique()
print(f"2) Unique status_ids: {num_unique_status_ids}")

### 3) How many video posts have been made for live selling?
num_video_posts = df[df["status_type"] == "video"].shape[0]
print(f"3) Number of video posts: {num_video_posts}")

### 4) How many likes have been given to the photo posted on 4/19/2018 at 22:26?
likes_on_photo = df[(df["status_published"] == "4/19/2018 22:26") & (df["status_type"] == "photo")]["num_likes"].values
print(f"4) Likes on the photo posted on 4/19/2018 at 22:26: {likes_on_photo[0] if len(likes_on_photo) > 0 else 'Not Found'}")

### 5) Total shares for the video post posted on 7/15/2017 at 10:52?
shares_on_video = df[(df["status_published"] == "7/15/2017 10:52") & (df["status_type"] == "video")]["num_shares"].values
print(f"5) Shares for the video posted on 7/15/2017 at 10:52: {shares_on_video[0] if len(shares_on_video) > 0 else 'Not Found'}")

### 6) What does K-means do?
print("6) K-means groups similar elements.")

### 7) Output of scatter plot code
print("7) The given code will produce 5 clusters with midpoints.")

### 8) Method to find optimal clusters
print("8) The Elbow method is used to decide the optimal number of clusters.")

### 9) What is K-means used for?
print("9) K-means is used for Clustering.")

### 10) Is hierarchical clustering used for exploration?
print("10) TRUE - Hierarchical clustering can be used for exploration.")

### 11) Function used for K-means
print("11) The function used is KMeans.")

### 12) How to train K-means?
print("12) The K-means model is trained using fit().")

### 13) What does random_state do?
print("13) Random state ensures reproducibility of results when code is run multiple times.")

### 14) How to check which data is placed under which cluster?
print("14) The attribute 'labels_' is used to check which data is placed under which cluster.")

### 15) Why do we perform standard scaling?
print("15) Standard scaling is done to standardize the data.")

### 16) Inter-cluster variance of K-means with 2 clusters

# Drop unnecessary columns
df.drop(columns=["status_id", "status_published"], inplace=True)

# Encode the status_type column using LabelEncoder
label_encoder = LabelEncoder()
df["status_type"] = label_encoder.fit_transform(df["status_type"])

df = df.dropna(axis=1, how='all')

# Fill remaining NaN values with the mean of each column
df.fillna(df.mean(), inplace=True)


# Standardize the data using Min-Max Scaler
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)

# Create a K-Means model for 2 clusters
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(scaled_data)

# Get the inter-cluster variance (inertia)
inertia = kmeans.inertia_
print(f"16) Inter-cluster variance: {inertia}")
```
