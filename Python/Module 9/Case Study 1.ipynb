```python
import pandas as pd
from surprise import Dataset, Reader, SVD
from surprise.model_selection import cross_validate

# Step 1: Load the datasets
ratings = pd.read_csv('ratings.csv')
movies = pd.read_csv('movies.csv')

# Step 2: Find the unique number of users and movies in the ‘ratings’ dataset
num_users = ratings['userId'].nunique()
num_movies = ratings['movieId'].nunique()
print(f"Unique users: {num_users}, Unique movies: {num_movies}")

# Step 3: Create a rating matrix for the ‘ratings’ dataset and store it in ‘Ratings’
rating_matrix = ratings.pivot(index='userId', columns='movieId', values='rating')
Rating = rating_matrix.fillna(0)

# Step 4: Load the ‘ratings’ dataset as SVD’s Dataset object and compute 3-fold cross-validation
reader = Reader(rating_scale=(0, 5))
data = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)
svd = SVD()
cross_validate(svd, data, n_jobs=-1, cv=3)

# Step 5: Find all the movies rated as 5 stars by user id ‘5’ and store it in ‘ratings_1’
ratings_1 = ratings[(ratings['userId'] == 5) & (ratings['rating'] == 5)]
print(ratings_1)

# Step 6: Create a shallow copy of the ‘movies’ dataset and store the result in ‘user_5’
user_5 = movies.copy()
print(user_5.head())

# Step 7: Train a recommender system using the SVD object and predict the ratings for user id ‘5’
trainset = data.build_full_trainset()
svd.fit(trainset)
user_id = 5
movie_ids = ratings['movieId'].unique()
predictions = [svd.predict(user_id, movie_id) for movie_id in movie_ids]
predicted_ratings = [(prediction.iid, prediction.est) for prediction in predictions]
predicted_ratings = sorted(predicted_ratings, key=lambda x: x[1], reverse=True)

# Step 8: Print the top 10 movie recommendations for user id ‘5’
top_10_recommendations = predicted_ratings[:10]
print("Top 10 movie recommendations for user 5:")
for movie_id, rating in top_10_recommendations:
    print(f"Movie ID: {movie_id}, Predicted Rating: {rating}")
```
