```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the dataset
df = pd.read_csv('cars.csv')  # Make sure your dataset path is correct

# Step 2: Extract relevant columns ('mpg', 'disp', 'hp')
car_features = df[['mpg', 'disp', 'hp']]

# Step 3: Build K-Means model with 3 clusters
kmeans_3 = KMeans(n_clusters=3, random_state=42)
df['Cluster_3'] = kmeans_3.fit_predict(car_features)

# Step 4: Extract observations for each cluster
cluster_0 = df[df['Cluster_3'] == 0]
cluster_1 = df[df['Cluster_3'] == 1]
cluster_2 = df[df['Cluster_3'] == 2]

# Display the clusters
print("Cluster 0 observations:")
print(cluster_0)
print("Cluster 1 observations:")
print(cluster_1)
print("Cluster 2 observations:")
print(cluster_2)

# Step 1: Build K-Means model with 5 clusters
kmeans_5 = KMeans(n_clusters=5, random_state=42)
df['Cluster_5'] = kmeans_5.fit_predict(car_features)

# Step 2: Extract observations for each cluster (5 clusters)
cluster_0_5 = df[df['Cluster_5'] == 0]
cluster_1_5 = df[df['Cluster_5'] == 1]
cluster_2_5 = df[df['Cluster_5'] == 2]
cluster_3_5 = df[df['Cluster_5'] == 3]
cluster_4_5 = df[df['Cluster_5'] == 4]

# Display the clusters
print("Cluster 0 (5 clusters) observations:")
print(cluster_0_5)
print("Cluster 1 (5 clusters) observations:")
print(cluster_1_5)
print("Cluster 2 (5 clusters) observations:")
print(cluster_2_5)
print("Cluster 3 (5 clusters) observations:")
print(cluster_3_5)
print("Cluster 4 (5 clusters) observations:")
print(cluster_4_5)


# Visualize the clusters using a scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='mpg', y='hp', hue='Cluster_3', palette='viridis', s=100, edgecolor='black')
plt.title('K-Means Clusters (3 clusters) based on MPG and Horsepower')
plt.xlabel('MPG (miles per gallon)')
plt.ylabel('Horsepower')
plt.legend(title='Cluster')
plt.show()

```
