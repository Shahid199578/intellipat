```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# Step 1: Create an np.array from the given data
data = np.array([[5, 3],
                 [10, 15],
                 [15, 12],
                 [24, 10],
                 [30, 30],
                 [85, 70],
                 [71, 80],
                 [60, 78],
                 [70, 55],
                 [80, 91]])

# Step 2: Generate a scatter plot for the data
plt.figure(figsize=(8,6))
plt.scatter(data[:, 0], data[:, 1], color='blue', marker='o')
plt.title('Scatter Plot of Given Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()

# Step 3: Plot dendrogram to visualize the clustering linkage
# Perform hierarchical/agglomerative clustering
linked = linkage(data, 'ward')

# Create and plot the dendrogram
plt.figure(figsize=(10, 7))
dendrogram(linked)
plt.title('Dendrogram for Agglomerative Clustering')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.show()
```
