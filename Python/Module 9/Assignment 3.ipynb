```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Step 1: Load the Iris data
iris = load_iris()
X = iris.data  # Extract the data property

# Step 2: Train an Agglomerative Clustering model
agg_clust = AgglomerativeClustering(n_clusters=3)  # Using 3 clusters for the Iris dataset
agg_clust.fit(X)

# Step 3: Plot dendrogram to visualize clustering linkage
# Perform hierarchical/agglomerative clustering
linked = linkage(X, 'ward')

# Create a dendrogram
plt.figure(figsize=(10, 7))
dendrogram(linked)
plt.title('Dendrogram for Agglomerative Clustering')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.show()
```
