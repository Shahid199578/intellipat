# Filter Records Based on Conditions
Fetch records satisfying the following conditions: Tenure > 50 and gender is 'Female', gender is 'Male' and SeniorCitizen is 0, TechSupport is 'Yes' and Churn is 'No', Contract type is 'Month-to-month' and Churn is 'Yes'.

```python
# Fetch records where Tenure > 50 and gender is 'Female'
tenure_gt_50_female = churn[(churn['tenure'] > 50) & (churn['gender'] == 'Female')]

# Fetch records where gender is 'Male' and SeniorCitizen is 0
male_non_senior = churn[(churn['gender'] == 'Male') & (churn['SeniorCitizen'] == 0)]

# Fetch records where TechSupport is 'Yes' and Churn is 'No'
tech_support_yes_churn_no = churn[(churn['TechSupport'] == 'Yes') & (churn['Churn'] == 'No')]

# Fetch records where Contract type is 'Month-to-month' and Churn is 'Yes'
month_to_month_churn_yes = churn[(churn['Contract'] == 'Month-to-month') & (churn['Churn'] == 'Yes')]

# Display the filtered records to verify
tenure_gt_50_female.head()
male_non_senior.head()
tech_support_yes_churn_no.head()
month_to_month_churn_yes.head()
```

# Count Male Senior Citizens with Tech Support
Use a for loop to calculate the number of customers that are getting tech support and are male senior citizens.

```python
# Count the number of male senior citizens who are getting tech support
male_senior_tech_support_count = 0

# Iterate through the DataFrame using a for loop
for index, row in churn.iterrows():
    if row['gender'] == 'Male' and row['SeniorCitizen'] == 1 and row['TechSupport'] == 'Yes':
        male_senior_tech_support_count += 1

# Display the count
male_senior_tech_support_count
```
