```python
import pandas as pd

# Load the dataset
churn = pd.read_csv('customer_churn.csv')

# Initialize a counter
count = 0

# Iterate through each row using a for loop
for index, row in churn.iterrows():
    if row['TechSupport'] == 'Yes' and row['gender'] == 'Male' and row['SeniorCitizen'] == 1:
        count += 1

# Print the result
print(f"Number of male senior citizens getting tech support: {count}")
```
