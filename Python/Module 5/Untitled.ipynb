```python
import pandas as pd

# Load dataset
df = pd.read_csv("housing.csv")

# Display first few records to understand the structure
print(df.head())
```

```python
costliest_zip = df.loc[df['price'].idxmax(), 'zipcode']
print("Costliest house is in ZIP code:", costliest_zip)
```

```python
grade_10_count = df[df['grade'] == 10].shape[0]
print("Number of houses with grade 10:", grade_10_count)
```

```python
null_values = df.isnull().sum().sum()
print("Total null values in the dataset:", null_values)

```

```python
customer_waterfront = df[df['id'] == 9126100861]['waterfront']
print("Customer 9126100861 has a waterfront:", "Yes" if customer_waterfront.any() else "No")
```

```python
houses_with_3_views = df[df['view'] == 3].shape[0]
print("Number of houses with 3 views:", houses_with_3_views)
```

```python
lowest_price = df['price'].min()
print("Lowest price of a house:", lowest_price)
```

```python
largest_sqft_living_zip = df.loc[df['sqft_living'].idxmax(), 'zipcode']
print("ZIP code with the biggest sqft_living:", largest_sqft_living_zip)
```

```python
import pandas as pd

# Load dataset
housing = pd.read_csv("housing.csv")

# Filter rows based on the given conditions
filtered_data = housing[(housing["bedrooms"]==(housing["view"]==3)) & (housing["waterfront"]==1)]

# Display the filtered data
print(filtered_data)

df_filtered = housing.loc[(housing["yr_built"] < 1980) & 
                     (housing['floors'] > 2) & 
                     (housing['bedrooms'] > 2)]
print(df_filtered)
```

```python
import pandas as pd

# Load dataset
housing = pd.read_csv("housing.csv")

# Summary statistics (including mean, std, min, max, etc.)
housing_summary = housing.describe()
print("Summary Statistics:\n", housing_summary)

# Mean and Standard Deviation
housing_mean = housing.mean()
housing_std = housing.std()

print("\nMean:\n", housing_mean)
print("\nStandard Deviation:\n", housing_std)
```
