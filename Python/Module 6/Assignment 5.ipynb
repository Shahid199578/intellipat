```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('cars.csv')

grouped_data = df.groupby('am')['carb'].sum()

# Generate the area chart
plt.figure(figsize=(10, 6))
plt.fill_between(grouped_data.index, grouped_data, color='skyblue', alpha=0.5)

# Labeling the axes and title
plt.xlabel('Transmission')
plt.ylabel('Number of carburetors')
plt.title('Transmission vs Number of carburetors')

# Show the plot
plt.show()
```
