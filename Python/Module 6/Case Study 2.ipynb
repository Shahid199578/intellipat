```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('cars.csv')  # Replace with the actual dataset filename

# Set the style for seaborn
sns.set_style("whitegrid")

# Bar Plot: Number of Cylinders vs Frequency of Cars
plt.figure(figsize=(8, 6))
df['cyl'].value_counts().sort_index().plot(kind='bar', color='skyblue', edgecolor='black')
plt.xlabel('Number of cylinders')
plt.ylabel('Frequency of cars')
plt.title('Cylinders vs Frequency of Cars')
plt.xticks(rotation=0)
plt.show()

# Bar Graph for Count of Null Values in Each Column
plt.figure(figsize=(8, 6))
df.isnull().sum().plot(kind='bar', color='red', edgecolor='black')
plt.xlabel('Columns')
plt.ylabel('Count of Null Values')
plt.title('Count of Null Values in Each Column')
plt.xticks(rotation=45)
plt.show()

# Histogram for 'mpg' (Miles Per Gallon)
plt.figure(figsize=(8, 6))
plt.hist(df['mpg'], bins=30, color='green', edgecolor='black')
plt.xlabel('Miles per gallon')
plt.ylabel('Frequency')
plt.title('Miles Per Gallon Histogram')
plt.show()

# Box Plot for 'hp' (Horsepower)
plt.figure(figsize=(8, 6))
sns.boxplot(x=df['hp'], color='purple')
plt.xlabel('Car Horsepower')
plt.title('Boxplot for Car Horsepower')
plt.show()
```
