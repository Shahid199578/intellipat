```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('cars.csv')

# Plot the pie chart
plt.figure(figsize=(8, 8))
cyl_counts = df['cyl'].value_counts()
plt.pie(cyl_counts, labels=cyl_counts.index, autopct='%1.1f%%', colors=['skyblue', 'lightcoral', 'lightgreen'])

# Set title
plt.title('Distribution of Cylinders in mtcars Dataset')

# Show the plot
plt.show()
```
