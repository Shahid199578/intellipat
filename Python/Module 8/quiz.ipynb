```python
import pandas as pd

# Load the data
data = pd.read_csv('data.csv')

# Filter employees with more than 5 years experience and salary more than 60000
filtered_data = data[(data['YearsExperience'] > 5) & (data['Salary'] > 60000)]

# Count the number of rows that satisfy the condition
employee_count = filtered_data.shape[0]
print(employee_count)
```

```python
# Filter employees earning between 50000 and 80000
salary_range_data = data[(data['Salary'] >= 50000) & (data['Salary'] <= 80000)]

# Count the number of rows
employee_count_in_range = salary_range_data.shape[0]
print(employee_count_in_range)
```

```python
import pandas as pd

# Load the data
housing_data = pd.read_csv('housing.csv')
# Filter houses with 2 floors
houses_with_2_floors = housing_data[housing_data['floors'] == 2]
print(f"Number of houses with 2 floors: {houses_with_2_floors.shape[0]}")


# Filter houses built before 1960 with a waterfront
houses_before_1960_with_waterfront = housing_data[(housing_data['yr_built'] < 1960) & (housing_data['waterfront'] == 1)]
print(f"Number of houses built before 1960 with a waterfront: {houses_before_1960_with_waterfront.shape[0]}")


# Filter houses with more than 4 bathrooms and find the most expensive one
expensive_house = housing_data[housing_data['bathrooms'] > 4]
most_expensive_house_price = expensive_house['price'].max()
print(f"Price of the most expensive house with more than 4 bathrooms: {most_expensive_house_price}")

```
