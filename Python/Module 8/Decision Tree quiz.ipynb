```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import precision_score, recall_score

# Load the dataset
df = pd.read_csv("Pokemon.csv")

# Replace blank values with NaN
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)

# Convert categorical columns to numerical using LabelEncoder
le = LabelEncoder()
for col in df.select_dtypes(include=['object']).columns:
    df[col] = le.fit_transform(df[col].astype(str))

# 3) Handle missing values (Fill NaN with column mean for numeric columns)
df.fillna(df.select_dtypes(include=[np.number]).mean(numeric_only=True), inplace=True)

# 1) How many Pokémon are from the 5th generation?
gen_5_count = df[df["Generation"] == 5].shape[0]
print(f"1) Number of Pokémon from Generation 5: {gen_5_count}")

# 2) How many Pokémon have the highest defense score?
max_defense = df["Defense"].max()
highest_defense_count = df[df["Defense"] == max_defense].shape[0]
print(f"2) Pokémon with highest defense: {highest_defense_count}")

# 4) Check correlation with Generation column
correlation = df.corr()["Generation"]
print(correlation)


# 5) Models for predicting Legendary status
X = df.drop(columns=["Legendary"])  # Features
y = df["Legendary"].astype(int)  # Target variable (convert True/False to 1/0)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Train Decision Tree model
model = DecisionTreeClassifier(random_state=1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# 6) Precision when target is False
precision = precision_score(y_test, y_pred, pos_label=0)
print(f"6) Precision when target is False: {precision:.2f}")

# 7) Sensitivity (Recall) when target is True
sensitivity = recall_score(y_test, y_pred, pos_label=1)
print(f"7) Sensitivity when target is True: {sensitivity:.2f}")

# 8) Misclassified data count
misclassified = sum(y_pred != y_test)
print(f"8) Misclassified data count: {misclassified}")

# 9) Avoiding bias in Decision Trees
print("9) Balance the dataset prior to fitting.")

# 10) Predicting number of oxygen tanks (Regression problem)
print("10) Decision Tree (best for numerical prediction).")

# 11) True statements about Decision Trees
print("11) Decision Trees can handle both categorical & numerical data, generate rules, and require low computation.")

# 12) Total gain computation
print("12) True - total gain = expected value - costs.")

# 13) Avoiding overfitting in Decision Trees
print("13) Best way to avoid overfitting: Pruning or stopping tree growth.")
```
