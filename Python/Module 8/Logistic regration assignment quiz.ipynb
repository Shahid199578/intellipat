```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, f1_score, classification_report

# Load dataset
df = pd.read_csv("census-income.csv")  # Ensure the correct file path

# Rename the last column
df.rename(columns={df.columns[-1]: "Annual Income"}, inplace=True)

# Convert missing values ('?') to NaN
df.replace("?", np.nan, inplace=True)

# 3. Count total missing values
total_missing_values = df.isnull().sum().sum()
print(f"3) Total missing values: {total_missing_values}")

# Drop missing values
df.dropna(inplace=True)

# 1. Count unique occupations (excluding NaN values)
num_occupations = df["occupation"].nunique()
print(f"1) Number of unique occupations: {num_occupations}")  # Answer: 14, 11, 15, or 13?

# 2. Count people in "Tech-support" with income >50K
tech_support_high_income = df[(df["occupation"] == "Tech-support") & (df["Annual Income"] == ">50K")].shape[0]
print(f"2) Tech Support making >50K: {tech_support_high_income}")



# 4. Replace missing values in 'Marital Status' using mode
df["marital-status"] = df["marital-status"].fillna(df["marital-status"].mode()[0])

print(f"4) Fill missing marital status with: Mode")

# 5. Count people with private workclass and not from the USA
private_non_usa = df[(df["workclass"] == "Private") & (df["native-country"] != "United-States")].shape[0]
print(f"5) Private workclass, non-USA: {private_non_usa}")

# 6. Count people with income <=50K or working ≥40 hrs/week
count_condition = df[(df["Annual Income"] == "<=50K") | (df["hours-per-week"] >= 40)].shape[0]
print(f"6) People with Income <=50K or working >=40 hrs: {count_condition}")

# 7. Methods for handling outliers
print(f"7) Handling Outliers: Both IQR & Z-score")

# 8. Chi-square usage
print(f"8) Chi-square is used to: Compare observed vs expected & check relationships")

# 9. VIF definition
print(f"9) VIF detects multicollinearity")

# 10. predict_proba in Logistic Regression
print(f"10) predict_proba gives class probabilities")

# 11. Logistic regression is used for classification
print(f"11) Logistic regression is NOT for regression problems")

# 12. Logit(0) transformation in logistic regression
print(f"12) Logit(0) transforms to probability: 0.5")

# 13. Logistic regression for categorical dependent variable with >2 values
print(f"13) Multinomial logistic regression")

### **Data Preprocessing for Model Training**


# Convert categorical variables into numerical using LabelEncoder
le = LabelEncoder()
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = le.fit_transform(df[col])


df = pd.get_dummies(df, drop_first=True)
# Split dataset into train (70%) and test (30%)
X = df.drop(columns=["Annual Income"])
y = df["Annual Income"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

# Train Logistic Regression Model
model = LogisticRegression(max_iter=500)
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# 14. Accuracy Score
accuracy = accuracy_score(y_test, y_pred)
print(f"14) Model Accuracy: {accuracy}")

# 15. Specificity (True Negative Rate)
tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
specificity = tn / (tn + fp)
print(f"15) Specificity: {specificity}")

# 16. Precision when target is False (<=50K)
precision_false = precision_score(y_test, y_pred, pos_label=0)
print(f"16) Precision for False class: {precision_false}")

# 17. Support value (Total test records)
support_value = len(y_test)
print(f"17) Total Support Value: {support_value}")

# 18. F1-score when target is True (>50K)
f1_true = f1_score(y_test, y_pred, pos_label=1)
print(f"18) F1-score for >50K: {f1_true}")

# 19. Correctly Classified Records
correctly_classified = (y_test == y_pred).sum()
print(f"19) Correctly classified records: {correctly_classified}")
```
