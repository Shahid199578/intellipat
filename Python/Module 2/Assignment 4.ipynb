<h2>Create three classes named A, B and C
<br>a. Inside the A class, create a constructor. Inside the constructor, initialize
2 global variables name and age.
<br>b. After initializing the global variables inside the constructor, now create a
function named ‘details’ and that function should return the ‘name’ variable.
<br>c. Inside the B class, create a constructor. Inside the constructor, initialize 2
global variables name and id.
<br>d. After initializing the global variables inside the constructor, now create a
function named ‘details’ and that function should return the ‘name’ variable.
<br>e. The C class should inherit from class A, and B. Inside the class C,
create a constructor, and inside the constructor, call the constructor of
class A.
<br>f. Now, create a method inside the class C, as get_details, and this function
should return the value of name.
<br>g. Atlast, create an object of class C, and with the help of the object, call
the get_details().
</h2>

```python
class A:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return self.name
class B:
    def __init__(self, name, id):
        self.name = name
        self.id = id
    def details(self):
        return self.name
class C(A, B):
    def __init__(self, name, age, id):
        A.__init__(self, name, age)
    def get_details(self):
        return self.name

obj = C("Mohd", 25, 1234)

obj.get_details()
```

<h2>Create a class named ‘Sub1’, inside the class, generate a user defined
function named ‘first’ and inside the function, pass the following statement
in the print()- ‘This is the first function from Sub 1 class’.
<br>a. Now create another class named ‘Sub2’, and inside the class, create a
function named ‘second’, and pass the following message in the print()-
‘This is the second function from the Sub 2 class’.
<br>b. After that, create another class named ‘Super’ and inside that class,
create a method named ‘final’, and pass the below message in the print()-
‘This is the final method from the super class’.
<br>c. Now, create an object for the Super class and call all the 3 user defined
methods, i.e., first(), second(), and final().

</h2>

```python
class Sub1:
    def first(self):
        print("This is the first function from Sub 1 class.")
class Sub2:
    def second(self):
        print('This is the second function from the Sub 2 class.')
class Super(Sub1, Sub2):
    def final(self):
        print('This is the final method from super class.')


obj = Super()

obj.first()
obj.second()
obj.final()
```

<h2>Create a class named ‘Parent’, and inside the class, create a function
named ‘fun1’ and pass the following message in the print()- ‘This is the

<br>a. Now create a class named ‘Child’ and inside the class, create a method
named ‘fun2’ and pass the following message in the print()- ‘This is the

<br>b. After that, create another class named ‘Hybrid’ and inside the class,
create a method named ‘fun3’ and pass the following message in the
print()- ‘This is the message from the fun3’.
<br>c. Now create an object of Hybrid class and with the help of the object, call

```python
class Parent:
    def fun1(self):
        print("This is the message from the fun1.")

class Clild(Parent):
    def fun2(self):
        print("this is the message from fun2.")

class Hybrid(Clild):
    def fun3(self):
        print("This is the message from fun3.")

obj = Hybrid()
obj.fun1()
obj.fun2()
obj.fun3()
```
