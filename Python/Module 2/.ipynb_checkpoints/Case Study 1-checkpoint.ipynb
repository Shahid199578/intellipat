<h2>1.Create a class named Employee, with a constructor ‘__init__’ method that
accepts name and salary as parameters and set properties named name
and salary.
</h2>

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
```

<h2>2. Define __str__ method in Employee class so that when someone tries to
print the object the string Name: employee_name, Salary:
employee_salary is printed with the actual employee name and salary.
</h2>

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def __str__(self):
        return f"Name: {self.name}, Salary: {self.salary}"


obj = Employee("Mohd", 35000)
print(obj)
```

<h2><br>3. Create another class named Calculator, with methods to add, subtract,
multiply and divide two numbers.
    <br>4. These methods take two numbers as parameters.
    <br>5. These methods will be called by a method named execute command. 
    <br>6. Execute command takes in 3 parameters command which is string that can be either ‘add’, ‘sub’, ‘mul’, ‘div’, and two numbers and it will call the appropriate method based on command parameter.¶
</h2>

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def __str__(self):
        return f"Name: {self.name}, Salary: {self.salary}"

class Calculator:

    def add(self, a,b):
        return a + b
    def sub(self, a,b):
        return a - b
    def mul(self, a,b):
        return a*b
    def divide(self, a,b):
        return a / b
    def execute_command(self, command, a, b):
        if command == 'add':
            return self.add(a,b)
        elif command == 'sub':
            return self.sub(a,b)
        elif command == 'mul':
            return self.mul(a,b)
        elif command == 'divide':
            return self.divide(a,b)
            
        


obj = Employee("Mohd", 35000)
calculator = Calculator()
print("Addition:", calculator.execute_command('add', 10,5))
print("Subtraction:", calculator.execute_command('sub', 10,5))
print("Multiplication:", calculator.execute_command('mul', 10,5))
print("Division:", calculator.execute_command('divide', 10,5))
print(obj)
```
