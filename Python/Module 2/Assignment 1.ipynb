<h2>1. Create a function named ‘factor’ that can only accept 1 argument. The
function should return the factorial of that number.
</h2>


```python
def factor(n):
    if n == 0:
        return 1
    else:
        return n * factor(n - 1)

num = int(input("Enter the number for check the factorial: "))
print(f"The factorial of {num} is {factor(num)}")


```

<h2>2. Create a function named ‘check_string’, the function should accept a string 
data from the user and the function should check if the user input contains
the letter ‘s’ in it. If it contains the letter ‘s’ then print- ‘The string is
containing the letter ‘s’, if not then print- ‘The string doesn’t contain the
letter ‘s’.
</h2>

```python
def check_string(s):
    if 's' in s:
        print ("The string is containing the letter ‘s’")
    else:
        print ("The string doesn’t contain the letter ‘s’")
user_String = input("Enter a string: ")
check_string(user_String)    
```

<h3>4. Create a lambda function that should double or multiply the number (that we will be passing in the lambda function) by 2.
Store the lambda function in a variable named ‘double_num’.</h3>


```python
double_num = lambda x: x * 2

num = int(input("Enter a number to double: "))
print(f"The double of {num} is {double_num(num)}")
```

<h3>5. Take user input string and check whether that string is palindrome or not</h3>

```python
def palindrome(s):
    return s == s[::-1]

user_string = input("Enter a string to check if it is a palindrome: ")
if palindrome(user_string):
    print(f"The string '{user_string}' is a palindrome.")
else:
    print(f"The string '{user_string}' is not a palindrome.")
```
