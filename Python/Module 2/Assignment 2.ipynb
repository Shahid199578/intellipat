<h2>1. Create a class named ‘Super’ and inside that class define a user-defined
function named fun1
<br>a. Inside the ‘fun1’ function, pass the message “This is function 1 in the
Super class.” in the print statement.</h2>

```python
class super:

    def fun1(self):
        print('This is function 1 in the Super class.')
obj = super()
obj.fun1()
```

<h2>3. Create 2 methods named ‘Hello’. In the 1st Hello method, pass only one
argument and pass this message: ‘This function is only having 1
argument’. And in the 2nd Hello method, pass two arguments and pass
this message: ‘This function is having 2 arguments’.
<br>a. Try to call both the methods and analyze the output of both the methods.</h2>

```python
class Method:

    def Hello(self, arg1=None, arg2=None):
        if arg1 is not None and arg2 is None:
            print("This function is only having 1 argument")
        elif arg1 is not None and arg2 is not None:
            print("This function is having 2 arguments")
        else:
            print("No arguments provided")

obj = Method()

obj.Hello("Arg1")
obj.Hello("Arg1", "arg2")
obj.Hello()
```

<h2>4. Create a method named ‘Sum’ that can accept multiple user inputs. Now
add those user defined input values using for loop and the function should
return the addition of the numbers.</h2>

```python
class Calculator:
    def Sum(self, *args):
        total = 0
        for num in args:
            total += num
        return total

calculator_instance = Calculator()

user_inputs = list(map(int, input("Enter numbers separated by spaces: ").split()))

result = calculator_instance.Sum(*user_inputs)


print("The sum of the numbers is:", result)
```

<h2>5. Create a class named ‘Encapsulation’:
<br>a. Inside the class, first create a constructor. Inside the constructor,
initialize original Value variable as 10.
<br>b. After creating the constructor, define a function named ‘Value’ and this
function should return the variable that we have initialized in the
constructor.
</h2>

```python
class Encapsulation:
    def __init__(self):
        self.__original_value = 10

    def Value(self):
        return self.__original_value

obj = Encapsulation()
print("The original value is:", obj.Value())
```
